```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import json
from datetime import datetime

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}


urls = [
    "https://www.liputan6.com/bisnis",
    "https://www.liputan6.com/tekno",
    "https://www.liputan6.com/news",
    "https://www.liputan6.com/showbiz",
    "https://www.liputan6.com/bola",
    "https://www.liputan6.com/otomotif",
    "https://www.liputan6.com/global",
    "https://www.liputan6.com/popular"
]


social_media_links = {
    "facebook": "https://www.facebook.com/liputan6online",
    "twitter_x": "https://twitter.com/liputan6dotcom",
    "instagram": "https://www.instagram.com/liputan6",
    "youtube": "https://www.youtube.com/c/liputan6dotcom",
    "tiktok": "https://www.tiktok.com/@liputan6.com"
}

collected_data = []

for url in urls:
    try:
        response = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(response.text, "html.parser")


        all_links = soup.find_all("a", href=True)
        articles = [a for a in all_links if "/read/" in a["href"]]

        print(f"Menemukan {len(articles)} kandidat artikel di: {url}")

        for a in articles:
            link = a["href"]
            title = a.get_text(strip=True)

            if title and len(title) > 10:
                if not link.startswith("http"):
                    link = "https://www.liputan6.com" + link

                collected_data.append({
                    "title": title,
                    "url": link,
                    "source_category": url.split("/")[-1],
                    "scraped_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                })
    except Exception as e:
        print(f"Gagal mengambil data dari {url}: {e}")

df = pd.DataFrame(collected_data).drop_duplicates(subset=["url"])

df.to_csv("post_liputan6_berita.csv", index=False)
df.to_json("post_liputan6_berita.json", orient="records", indent=4)
df.to_excel("post_liputan6_berita.xlsx", index=False)

with open("social_media.json", "w") as f:
    json.dump(social_media_links, f, indent=4)

print(f"\nBerhasil menyimpan {len(df)} artikel unik ke CSV, JSON, dan XLSX.")
print("Tautan media sosial berhasil disimpan ke social_media.json.")
```
